# Задача 1. Stack vs queue

**Условие:** Реализовать стек и очередь на основе связных списков. Без использования сторонних библиотек. 

## 1. Решение

````python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class Stack:
    def __init__(self):
        self.top = None

    def is_empty(self):
        return self.top is None

    def push(self, value):
        new_node = Node(value)
        new_node.next = self.top
        self.top = new_node

    def pop(self):
        if self.is_empty():
            print("Стек пуст")
            return None

        value = self.top.value
        self.top = self.top.next
        return value

    def peek(self):
        if self.is_empty():
            print("Стек пуст")
            return None

        return self.top.value

    def print_stack(self):
        current = self.top

        print("Стек:", end=" ")
        while current is not None:
            print(current.value, end=" ")
            current = current.next
        print()

class Queue:
    def __init__(self):
        self.front = None
        self.rear = None

    def is_empty(self):
        return self.front is None

    def enqueue(self, value):
        new_node = Node(value)

        if self.is_empty():
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node

    def dequeue(self):
        if self.is_empty():
            print("Очередь пуста")
            return None

        value = self.front.value
        self.front = self.front.next

        if self.front is None:
            self.rear = None

        return value

    def peek(self):
        if self.is_empty():
            print("Очередь пуста")
            return None

        return self.front.value

    def print_queue(self):
        current = self.front

        print("Очередь:", end=" ")
        while current is not None:
            print(current.value, end=" ")
            current = current.next
        print()
````

## 2. Объяснение решения

В стеке достаточно хранить ссылку (top) на верхний элемент. При добавлении нового элемента (push) создаётся новый узел. Его поле next указывает на текущую вершину стека, после чего новый узел становится новой вершиной.
При выполнении pop() значение верхнего элемента сохраняется, а top перемещается на следующий узел.
Таким образом реализуется принцип LIFO.

В очереди используются две ссылки:
front — первый элемент очереди, rear — последний элемент очереди.
При добавлении элемента методом enqueue новый узел присоединяется после rear. При удалении методом dequeue удаляется элемент, на который указывает front.
Благодаря наличию ссылки rear новый элемент можно добавить в конец очереди за постоянное время.

## 3. Оценка сложности

Пусть n — количество элементов в структуре.

### Стек

| Операция     | Сложность |
| ------------ | --------- |
| `push()`     | O(1)      |
| `pop()`      | O(1)      |
| `peek()`     | O(1)      |
| `is_empty()` | O(1)      |
| `len()`      | O(1)      |
| `to_list()`  | O(n)      |

Добавление и удаление выполняются только в начале связного списка, поэтому обход списка не требуется.

Память:

O(n), поскольку для каждого из n элементов создаётся отдельный узел связного списка.

### Очередь

| Операция     | Сложность |
| ------------ | --------- |
| `enqueue()`  | O(1)      |
| `dequeue()`  | O(1)      |
| `peek()`     | O(1)      |
| `is_empty()` | O(1)      |
| `len()`      | O(1)      |
| `to_list()`  | O(n)      |

Добавление выполняется через указатель rear, а удаление — через front, поэтому обход списка не требуется.

Память: O(n).

## 4. Пошаговая визуализация

Рассмотрим последовательность операций:

```python
stack.push(5)
stack.push(10)
stack.push(15)
stack.pop()
stack.push(20)
```

Состояния стека:

| Шаг | Операция   | Состояние стека |
| --- | ---------- | --------------- |
| 1   | `push(5)`  | `5`             |
| 2   | `push(10)` | `10 → 5`        |
| 3   | `push(15)` | `15 → 10 → 5`   |
| 4   | `pop()`    | `10 → 5`        |
| 5   | `push(20)` | `20 → 10 → 5`   |

Вершина находится слева.

Теперь рассмотрим очередь:

```python
queue.enqueue(5)
queue.enqueue(10)
queue.enqueue(15)
queue.dequeue()
queue.enqueue(20)
```

| Шаг | Операция      | Состояние очереди |
| --- | ------------- | ----------------- |
| 1   | `enqueue(5)`  | `5`               |
| 2   | `enqueue(10)` | `5 → 10`          |
| 3   | `enqueue(15)` | `5 → 10 → 15`     |
| 4   | `dequeue()`   | `10 → 15`         |
| 5   | `enqueue(20)` | `10 → 15 → 20`    |

Вершина находится справа.

## 5. Тесты

````python
def validate_stack_sequences(pushed, popped):
    stack = []
    j = 0

    for value in pushed:
        stack.append(value)

        while (
            stack
            and j < len(popped)
            and stack[-1] == popped[j]
        ):
            stack.pop()
            j += 1

    return j == len(popped)


def test_validate_stack_sequences():
    assert validate_stack_sequences(
        [1, 2, 3, 4, 5],
        [4, 5, 3, 2, 1]
    ) is True

    assert validate_stack_sequences(
        [1, 2, 3, 4, 5],
        [4, 3, 5, 1, 2]
    ) is False

    assert validate_stack_sequences(
        [1, 2, 3, 4],
        [1, 2, 3, 4]
    ) is True

    assert validate_stack_sequences(
        [1, 2, 3, 4],
        [4, 3, 2, 1]
    ) is True

    assert validate_stack_sequences(
        [10],
        [10]
    ) is True

    assert validate_stack_sequences(
        [],
        []
    ) is True

    assert validate_stack_sequences(
        [1, 2, 3],
        [3, 1, 2]
    ) is False

    assert validate_stack_sequences(
        [1, 2, 3, 4, 5, 6],
        [2, 1, 4, 3, 6, 5]
    ) is True

    assert validate_stack_sequences(
        [1, 2, 3, 4, 5, 6],
        [3, 2, 1, 5, 6, 4]
    ) is True

    assert validate_stack_sequences(
        [1, 2, 3, 4, 5, 6],
        [3, 2, 1, 6, 4, 5]
    ) is False

    assert validate_stack_sequences(
        [-3, -2, -1],
        [-1, -2, -3]
    ) is True

    assert validate_stack_sequences(
        [100, 200, 300, 400],
        [300, 400, 200, 100]
    ) is True

    print("Все тесты успешно пройдены")


test_validate_stack_sequences()
````
