# Задача 2. Anagrams

**Условие:** дан список слов. Сгруппируйте слова так, чтобы в одной группе оказались все анаграммы.

## 1. Решение

````python
def group_anagrams(strs: list[str]) -> list[list[str]]:
    i = 0
    dictionary = {}

    while i < len(strs):
        key = "".join(sorted(strs[i]))
        if key in dictionary:
            dictionary[key].append(strs[i])
        else:
            dictionary[key] = [strs[i]]
        i += 1

    return list(dictionary.values())
````

## 2. Объяснение решения

Идём по списку слов. Для каждого слова сортируем буквы и соединяем их обратно в строку. Полученная строка будет ключом в словаре `dictionary`.

Например, для `eat`, `tea` и `ate` получается один ключ — `aet`.

1. Если ключ уже есть в словаре, добавляем слово в готовый список.
2. Если ключа ещё нет, создаём для него список с текущим словом.
3. Когда все слова просмотрены, возвращаем списки из словаря.

У анаграмм после сортировки получается одинаковая строка, поэтому они попадут в одну группу. У слов с разными буквами или разным количеством повторений ключи отличаются, поэтому такие слова останутся в разных группах.

## 3. Оценка сложности

Пусть **n** — количество слов, **m** — максимальная длина слова.

- **Время: O(n · m log m)** при m ≥ 2. Для каждого слова сортируем до m букв за O(m log m), затем создаём строку и ищем её в словаре за O(m) в среднем с учётом вычисления хеша.
- **Дополнительная память: O(n · m)** при m ≥ 1. Словарь хранит ключи длиной до m и списки ссылок на исходные слова.

Для пустых и однобуквенных слов обработка списка занимает O(n). 


## 4. Пошаговая визуализация

Рассмотрим `strs = ["aab", "abb", "aba", "ab", "bba", "aab", "ba"]`. Здесь слова состоят из одних и тех же букв, но количество повторений разное. Кроме того, слово `aab` встречается дважды, а слова одной группы идут не подряд.

Перед циклом словарь пуст. На каждом шаге сортируем буквы текущего слова и по полученному ключу находим нужную группу.

| Индекс i | Слово | Ключ | Действие | Группа после добавления |
|---|---|---|---|---|
| 0 | aab | aab | Ключа нет — создаём группу | `["aab"]` |
| 1 | abb | abb | Другой ключ — создаём новую группу | `["abb"]` |
| 2 | aba | aab | Ключ aab уже есть — добавляем слово | `["aab", "aba"]` |
| 3 | ab | ab | Ключа ab ещё нет — создаём группу | `["ab"]` |
| 4 | bba | abb | Добавляем в группу abb | `["abb", "bba"]` |
| 5 | aab | aab | Добавляем повторное вхождение слова | `["aab", "aba", "aab"]` |
| 6 | ba | ab | Добавляем в группу ab | `["ab", "ba"]` |

После цикла словарь имеет вид:

```python
{
    "aab": ["aab", "aba", "aab"],
    "abb": ["abb", "bba"],
    "ab": ["ab", "ba"]
}
```

Возвращаем его значения: `[["aab", "aba", "aab"], ["abb", "bba"], ["ab", "ba"]]`.

Слова `aab`, `abb` и `ab` попали в разные группы: набор букв одинаковый, но количество каждой буквы отличается. Оба вхождения `aab` сохранились — каждое слово из исходного списка добавляется в результат.


## 5. Тесты

````python
import unittest


def normalize(groups):
    return sorted(sorted(group) for group in groups)


class TestGroupAnagrams(unittest.TestCase):
    def test_example(self):
        result = group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"])
        expected = [["bat"], ["nat", "tan"], ["ate", "eat", "tea"]]
        self.assertEqual(normalize(result), normalize(expected))

    def test_empty_list(self):
        self.assertEqual(group_anagrams([]), [])

    def test_one_word(self):
        self.assertEqual(group_anagrams(["cat"]), [["cat"]])

    def test_all_anagrams(self):
        result = group_anagrams(["abc", "bca", "cab", "acb"])
        expected = [["abc", "bca", "cab", "acb"]]
        self.assertEqual(normalize(result), normalize(expected))

    def test_no_anagrams(self):
        result = group_anagrams(["cat", "dog", "bird"])
        expected = [["cat"], ["dog"], ["bird"]]
        self.assertEqual(normalize(result), normalize(expected))

    def test_duplicate_words(self):
        result = group_anagrams(["eat", "eat", "tea", "bat", "bat"])
        expected = [["eat", "eat", "tea"], ["bat", "bat"]]
        self.assertEqual(normalize(result), normalize(expected))

    def test_repeated_letters(self):
        result = group_anagrams(["aab", "aba", "abb", "bba", "ab", "ba"])
        expected = [["aab", "aba"], ["abb", "bba"], ["ab", "ba"]]
        self.assertEqual(normalize(result), normalize(expected))

    def test_empty_strings(self):
        result = group_anagrams(["", "a", "", "aa"])
        expected = [["", ""], ["a"], ["aa"]]
        self.assertEqual(normalize(result), normalize(expected))

    def test_single_letters(self):
        result = group_anagrams(["b", "a", "b", "c", "a"])
        expected = [["b", "b"], ["a", "a"], ["c"]]
        self.assertEqual(normalize(result), normalize(expected))


suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestGroupAnagrams)
result = unittest.TextTestRunner(verbosity=2).run(suite)
assert result.wasSuccessful(), "Есть непройденные тесты"
````
