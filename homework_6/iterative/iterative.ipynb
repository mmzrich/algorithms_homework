# Задача 2. Iterative

**Объяснение решения**

`merge_sort` начинает с блоков длины `1`: они уже отсортированы. Сливаем соседние блоки, каждый раз выбирая меньший из первых оставшихся элементов. После прохода удваиваем `width`. Так получаем отсортированные блоки длины `2`, `4`, `8` и далее, пока весь массив не станет одним блоком.

`quick_sort` хранит в стеке пары `(left, right)` — границы еще не отсортированных участков, включая оба конца. Берем участок и выбираем последний элемент опорным. За один проход переносим значения `<= pivot` в начало участка, затем ставим опорный сразу после них. Он оказывается на своем месте. В стек добавляем левую и правую части, если в них хотя бы два элемента.

Обе функции работают циклами и возвращают новый список. Исходный массив не меняется.

**Решение**

````python
def merge_sort(nums):
    result = nums.copy()
    n = len(result)
    width = 1

    while width < n:
        for start in range(0, n, 2 * width):
            left = result[start:start + width]
            right = result[start + width:start + 2 * width]
            i = j = 0

            for k in range(start, min(start + 2 * width, n)):
                if i < len(left) and (j == len(right) or left[i] <= right[j]):
                    result[k] = left[i]
                    i += 1
                else:
                    result[k] = right[j]
                    j += 1

        width *= 2

    return result
````

````python
def quick_sort(nums):
    result = nums.copy()
    stack = [(0, len(result) - 1)]

    while stack:
        left, right = stack.pop()
        if left >= right:
            continue

        pivot = result[right]
        i = left

        for j in range(left, right):
            if result[j] <= pivot:
                result[i], result[j] = result[j], result[i]
                i += 1

        result[i], result[right] = result[right], result[i]

        if left < i - 1:
            stack.append((left, i - 1))
        if i + 1 < right:
            stack.append((i + 1, right))

    return result
````

**Пример**

Возьмем `[4, -2, 7, 4, 0, 9, -5, 2, 7, 1]`. Здесь есть отрицательные числа, ноль и повторы. Длина не является степенью двойки, поэтому при слиянии будут неполные блоки.

````python
nums = [4, -2, 7, 4, 0, 9, -5, 2, 7, 1]

print("Исходный массив:", nums)
print("Merge sort:", merge_sort(nums))
print("Quick sort:", quick_sort(nums))
````

**Merge sort**

`width` — длина блока до слияния. В каждой строке показан массив после полного прохода. Квадратными скобками выделены отсортированные блоки.

| Проход | width | Блоки после слияния |
|---|---|---|
| 1 | `1` | `[-2, 4] [4, 7] [0, 9] [-5, 2] [1, 7]` |
| 2 | `2` | `[-2, 4, 4, 7] [-5, 0, 2, 9] [1, 7]` |
| 3 | `4` | `[-5, -2, 0, 2, 4, 4, 7, 9] [1, 7]` |
| 4 | `8` | `[-5, -2, 0, 1, 2, 4, 4, 7, 7, 9]` |

Например, при `width = 2` блоки `[0, 9]` и `[-5, 2]` сливаются так:

| Шаг | Первые оставшиеся элементы | Выбираем | Результат |
|---|---|---|---|
| 1 | `0` и `-5` | `-5` справа | `[-5]` |
| 2 | `0` и `2` | `0` слева | `[-5, 0]` |
| 3 | `9` и `2` | `2` справа | `[-5, 0, 2]` |
| 4 | `9`, правая часть пуста | `9` слева | `[-5, 0, 2, 9]` |

У последнего блока `[1, 7]` на проходах с `width = 2` и `4` нет соседа. Правая часть пустая, поэтому левая просто переписывается.

**Quick sort**

Первый опорный — `1`. Переносим `-2`, `0` и `-5` в начало участка. Индекс `i` становится равен `3`. Меняем местами опорный и элемент на позиции `i`:

```text
[4, -2, 7, 4, 0, 9, -5, 2, 7, 1]
                    ↓
[-2, 0, -5] + [1] + [4, 9, 7, 2, 7, 4]
```

Теперь `1` стоит на своем месте. В стек кладем `(0, 2)` и `(4, 9)`. Верх стека справа, поэтому правую часть обрабатываем первой.

| Шаг | Участок | Опорный | Массив после разбиения | Стек после шага |
|---|---|---|---|---|
| 1 | `(0, 9)` | `1` | `[-2, 0, -5, 1, 4, 9, 7, 2, 7, 4]` | `[(0, 2), (4, 9)]` |
| 2 | `(4, 9)` | `4` | `[-2, 0, -5, 1, 4, 2, 4, 9, 7, 7]` | `[(0, 2), (4, 5), (7, 9)]` |
| 3 | `(7, 9)` | `7` | `[-2, 0, -5, 1, 4, 2, 4, 7, 7, 9]` | `[(0, 2), (4, 5)]` |
| 4 | `(4, 5)` | `2` | `[-2, 0, -5, 1, 2, 4, 4, 7, 7, 9]` | `[(0, 2)]` |
| 5 | `(0, 2)` | `-5` | `[-5, 0, -2, 1, 2, 4, 4, 7, 7, 9]` | `[(1, 2)]` |
| 6 | `(1, 2)` | `-2` | `[-5, -2, 0, 1, 2, 4, 4, 7, 7, 9]` | `[]` |

Стек пуст — все участки обработаны.

**Сложность**

Пусть `n` — длина массива, сравнение чисел занимает `O(1)`.

| Алгоритм | Время | Дополнительная память |
|---|---|---|
| Merge sort | `O(n log n)` | `O(n)` |
| Quick sort, в среднем на случайных различных числах | `O(n log n)` | `O(n)` |
| Quick sort, худший случай | `O(n²)` | `O(n)` |

У merge sort `O(log n)` проходов, на каждом обрабатываем `O(n)` элементов. Копия массива и временные части для слияния занимают `O(n)` памяти.

У quick sort одно разбиение линейно по длине участка. При равномерных разбиениях получаем `O(n log n)`. На упорядоченных массивах и одинаковых числах выбранный опорный каждый раз оставляет часть длины `n - 1`, поэтому время — `O(n²)`.

Копия массива занимает `O(n)`. Стек quick sort в худшем случае тоже занимает `O(n)`.

**Тесты**

````python
import random
import unittest
from itertools import product


class TestIterativeSort(unittest.TestCase):
    def check(self, nums):
        before = nums.copy()
        expected = sorted(nums)

        for function in (merge_sort, quick_sort):
            with self.subTest(function=function.__name__, size=len(nums)):
                result = function(nums)
                self.assertEqual(result, expected)
                self.assertEqual(nums, before)
                self.assertIsNot(result, nums)

    def test_empty(self):
        self.check([])

    def test_single(self):
        self.check([7])

    def test_two_elements(self):
        for nums in ([1, 2], [2, 1], [2, 2]):
            self.check(nums)

    def test_example(self):
        self.check([4, -2, 7, 4, 0, 9, -5, 2, 7, 1])

    def test_duplicates(self):
        self.check([3, 1, 3, -2, 1, 3, -2, 0])

    def test_all_equal(self):
        self.check([5] * 2000)

    def test_sorted(self):
        self.check(list(range(2000)))

    def test_reversed(self):
        self.check(list(range(1999, -1, -1)))

    def test_negative_and_zero(self):
        self.check([0, -9, -1, 0, -5, -9])

    def test_large_numbers(self):
        self.check([10**100, 0, -10**100, 7, -7])

    def test_floats(self):
        self.check([1.5, -0.5, 2.0, 0.0, 1.5])

    def test_different_lengths(self):
        rng = random.Random(42)
        for size in (3, 5, 7, 8, 9, 15, 16, 17, 31, 32, 33, 127, 128, 129):
            self.check(rng.sample(range(500), size))

    def test_random(self):
        rng = random.Random(42)
        for size in (10, 100, 1000, 5000):
            self.check([rng.randint(-1000, 1000) for _ in range(size)])

    def test_all_small_arrays(self):
        for size in range(7):
            for values in product((-1, 0, 1), repeat=size):
                self.check(list(values))

    def test_independent_results(self):
        for function in (merge_sort, quick_sort):
            nums = [3, 1, 2]
            first = function(nums)
            first[0] = 99
            self.assertEqual(nums, [3, 1, 2])
            self.assertEqual(function(nums), [1, 2, 3])
````

````python
suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestIterativeSort)
test_result = unittest.TextTestRunner(verbosity=2).run(suite)
assert test_result.wasSuccessful(), "Некоторые тесты завершились ошибкой"
print(f"Успешно пройдено тестов: {test_result.testsRun}")
````
