# Задача 3. Merge lists

**Условие:** Даны два отсортированных односвязных списка list1 и list2

Необходимо объединить их в один новый отсортированный список.
Новый список должен быть составлен слиянием узлов двух исходных списков.
Вернуть необходимо голову объединенного списка.

## 1. Решение

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


def create_list(values):
    if not values:
        return None

    head = Node(values[0])
    current = head

    for value in values[1:]:
        current.next = Node(value)
        current = current.next

    return head


def to_list(head):
    result = []

    while head is not None:
        result.append(head.value)
        head = head.next

    return result


def merge_lists(list1, list2):
    dummy = Node(0)
    current = dummy

    while list1 is not None and list2 is not None:
        if list1.value <= list2.value:
            current.next = list1
            list1 = list1.next
        else:
            current.next = list2
            list2 = list2.next

        current = current.next

    if list1 is not None:
        current.next = list1
    else:
        current.next = list2

    return dummy.next
```

## 2. Объяснение решения

Есть два уже отсортированных односвязных списка. Сравниваем текущие элементы list1 и list2. Узел с меньшим значением добавляем в результирующий список, после чего переходим к следующему узлу соответствующего списка.
Когда один из списков заканчивается, оставшуюся часть второго списка сразу присоединяем к результату, так как она уже отсортирована. Новые узлы для результата не создаются, используются узлы исходных списков.

## 3. Оценка сложности

Пусть n — длина list1, m — длина list2.
Временная сложность: O(n + m) Каждый узел обоих списков обрабатывается не более одного раза.

Дополнительная память: O(1)
Создается только один новый узел dummy.

## 5. Тесты

```python
def test_merge_lists():
    list1 = create_list([1, 2, 4])
    list2 = create_list([1, 3, 4])

    result = merge_lists(list1, list2)
    assert to_list(result) == [1, 1, 2, 3, 4, 4]

    list1 = create_list([])
    list2 = create_list([1, 2, 3])

    result = merge_lists(list1, list2)
    assert to_list(result) == [1, 2, 3]

    list1 = create_list([1, 2, 3])
    list2 = create_list([])

    result = merge_lists(list1, list2)
    assert to_list(result) == [1, 2, 3]

    list1 = create_list([])
    list2 = create_list([])

    result = merge_lists(list1, list2)
    assert to_list(result) == []

    list1 = create_list([1, 5])
    list2 = create_list([2, 3, 4, 6, 7])

    result = merge_lists(list1, list2)
    assert to_list(result) == [1, 2, 3, 4, 5, 6, 7]

    list1 = create_list([-10, -3, 5])
    list2 = create_list([-7, 0, 8])

    result = merge_lists(list1, list2)
    assert to_list(result) == [-10, -7, -3, 0, 5, 8]

    print("Все тесты успешно пройдены")


test_merge_lists()
```
